## Prüfablauf

1. In „Bauteile“ die innere Geschoßdecke auswählen: Sie liegt innerhalb der Hülle.
2. „Thermische Hülle“ aktivieren: Die Kontur umschließt beide beheizten Geschosse.
3. „Wärmeflüsse“ aktivieren: ungekühlt zeigt Nachtlüftung, aber keine aktive Kühlung.
4. Auf „Mit aktiver Kühlung“ wechseln: Kühlung sichtbar, Nachtlüftung ausgeblendet.
5. Wärmeflüsse ausblenden und Hülle separat zuschalten.
6. Mit Tab und Eingabetaste bedienen; Desktop und schmale Browseransicht prüfen.

Die Pfeilrichtungen sind qualitative Beispiele. Die Pfeildicke kodiert keine
Leistung. Reale zeitabhängige Vorzeichen und Zahlen folgen erst mit den passenden
Zeitreihen. Die SVG-Datei ist die neutrale statische Grundlage; der interaktiv gewählte Zustand lässt sich über „Ansicht als SVG speichern“ exportieren.
