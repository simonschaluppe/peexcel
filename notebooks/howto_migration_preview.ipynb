# Migration-Anleitung als Vorschau prüfen

Die Anleitung in `website/migration.qmd` beschreibt den vollständigen Bedienablauf: aktuellen Stand speichern, Varianten auswählen und als XLSX exportieren, in eine neue Arbeitsmappe importieren, Eingaben prüfen und neu berechnen. Sieben bereitgestellte Originalaufnahmen zeigen Menüband, Exportauswahl, Importauswahl für Projektvarianten und PV-Profile sowie die Abschlussmeldung und die anschließende Simulationsfrage. Ein einzelner grüner Orientierungshinweis mit Icon nennt zu Beginn das Menüband und die Gruppe. Die ausführlichen Arbeitsschritte stehen im normalen Text; Klickbefehle verwenden die vorhandene Schaltflächenformatierung. Die Originalaufnahmen liegen gemeinsam in `website/illustrations/migration/`. Export und Import verwenden unterschiedliche Beispielprojekte. Die genaue Toolversion ist nicht sichtbar.

Dieses Notebook verwendet denselben Quarto-Aufruf wie die lokale Website-Vorschau; es implementiert keinen zweiten Import oder Export. Voraussetzung: Quarto im PATH. Die Excel-Schritte werden manuell im Tool ausgeführt; das Notebook behauptet keinen automatisierten Excel-Funktionstest.

Nach dem Rendern die HTML-Seite lokal öffnen oder über einen lokalen Webserver anzeigen. Menüband, Variantenauswahl, Importbefehl, Bildvergrößerung und Quellenfußnote prüfen. Hypothesis-Kommentare bleiben bis zur ausdrücklichen Abnahme erhalten. Erst danach die sieben Annotationen des übergebenen Exports mit berechtigtem Zugriff löschen. Dieses Notebook verändert keine Annotationen und veröffentlicht die Website nicht.


```python
from pathlib import Path
import subprocess

root = Path.cwd()
if not (root / "website/migration.qmd").exists():
    root = root.parent
assert (root / "website/migration.qmd").is_file()
for name in ("varianten", "export", "import", "import-auswahl", "import-pv", "import-fertig", "simulation"):
    assert (root / f"website/illustrations/migration/{name}.png").is_file()
subprocess.run(["quarto", "render", "website/migration.qmd"], cwd=root, check=True)
preview = root / "website/_site/migration.html"
html = preview.read_text(encoding="utf-8")
assert all(f"migration/{name}.png" in html for name in ("varianten", "export", "import", "import-auswahl", "import-pv", "import-fertig", "simulation"))
assert 'role="doc-endnotes"' in html
print(preview)
```
