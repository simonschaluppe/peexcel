# Quartiere bewerten: Energie und Klimawirkung

Ein Quartier lässt sich aus zwei ergänzenden Perspektiven betrachten: Wie stehen Energiebedarf und bewertete Deckung zueinander? Und welche Treibhausgasemissionen entstehen über den Lebenszyklus? Die beiden Grafiken machen diese Zusammenhänge sichtbar.

### 01 · Primärenergie

Bedarf und Deckung stehen sich gegenüber. Die Bewertung berücksichtigt neben Energieflüssen auch die Rahmenbedingungen des Quartiers.

### 02 · Treibhausgase

Herstellung und Betrieb tragen zur Klimawirkung bei. Die Gesamtbilanz wird einem Referenzbudget gegenübergestellt.

## Primärenergie: Bedarf und Deckung bewerten

Die PE-Bilanz stellt den bewerteten Bedarf der bewerteten Deckung gegenüber. Neben den Energieanwendungen und der lokalen erneuerbaren Versorgung berücksichtigt die Bewertungsmethodik auch Energieflexibilität und Kontextfaktoren. Die Werte beziehen sich auf ein Jahr und einen Quadratmeter Nettogrundfläche.

### So liest du die PE-Grafik

- **Bedarf:** Die linke Säule zeigt die bewerteten Beiträge der Energieanwendungen und gegebenenfalls negative Kontextbeiträge.
- **Deckung:** Die rechte Säule zeigt lokale Erneuerbare, Energieflexibilität und positive Kontextbeiträge.
- **Differenz:** Die Klammer zeigt Deckung minus Bedarf. Grün bedeutet eine positive PE-Bilanz, Rot eine negative.
- **Einzelbeiträge:** Farben und Schraffuren unterscheiden die Komponenten. Beim Zeigen auf einen Beitrag erscheint sein Wert.

### Was bedeutet eine positive PE-Bilanz?

Eine positive Differenz zwischen bewerteter Deckung und bewertetem Bedarf bedeutet nach der Bewertungsmethodik eine **positive PE-Bilanz** und damit ein **Plus-Energie-Quartier** innerhalb der betrachteten Bilanzgrenze: **im Betrieb** oder **mit Mobilität**, wenn die zugeordnete Mobilität einbezogen ist. Die Lebenszyklus-Klimawirkung wird ergänzend in der THG-Bilanz betrachtet.

### Kontextfaktoren: Rahmenbedingungen berücksichtigen

**Bauliche Dichte.** Die Geschossflächenzahl (GFZ) beschreibt das Verhältnis von Geschossfläche zu Grundstücksfläche. Der Dichtefaktor berücksichtigt diese Rahmenbedingung in der Bewertung. Bei geringer GFZ kann er **negativ** sein: Der Beitrag erscheint dann auf der Bedarfsseite und verschlechtert die PE-Bilanz. Das lässt sich an den Einfamilienhausbeispielen nachvollziehen. Ein positiver Dichtebeitrag steht dagegen auf der Deckungsseite.

**Mobilität.** Der Mobilitätskontext berücksichtigt die in der Bewertungsmethodik angesetzten Rahmenbedingungen der Mobilität. Er ist von den ebenfalls dargestellten Energieaufwendungen für Mobilität zu unterscheiden.

**Sanierung.** Der Sanierungskontext berücksichtigt die Rahmenbedingungen des Gebäudebestands bei der Bewertung. Sein ausgewiesener Beitrag ist als Bewertungskomponente zu lesen.

Kontextfaktoren sind keine zusätzliche Strom- oder Wärmeerzeugung. Sie verändern die bewertete Bilanz. Für den Vergleich lohnt sich deshalb der Blick darauf, welche Unterschiede aus der Energieversorgung und welche aus den Kontextbeiträgen entstehen.

## Wie liest man die Treibhausgasbilanz?

Die Lebenszyklusbetrachtung umfasst mehr als den jährlichen Energieverbrauch. **EE** steht hier für graue, insbesondere herstellungsbezogene Emissionen; **OE** für betriebliche Emissionen. Die farbigen Beiträge zeigen unter anderem Konstruktion, technische Anlagen, Energieversorgung und Mobilität. Gutschriften oder andere negative Beiträge stehen unterhalb der Nulllinie.

Die **Raute** zeigt die vorhandene Gesamtbilanz einschließlich negativer Beiträge. Die rote Linie markiert das hier verwendete **Referenzbudget von 320 kg CO₂e je Quadratmeter Bruttogrundfläche**. Die Beschriftung zeigt, wie weit der Gesamtwert darüber oder darunter liegt. Ein niedrigerer Wert bedeutet eine geringere bilanzierte Klimawirkung innerhalb der betrachteten Systemgrenze.

### So liest du die THG-Grafik

- **EE und OE:** Die Säulen unterscheiden graue, insbesondere herstellungsbezogene Emissionen und betriebliche Emissionen.
- **Negative Beiträge:** Gutschriften und andere negative Beiträge stehen unterhalb der Nulllinie und gehen mit ihrem Vorzeichen in die Gesamtbilanz ein.
- **Raute:** Sie zeigt die gesamte bilanzierte Klimawirkung.
- **Rote Linie:** Sie markiert das Referenzbudget von 320 kg CO₂e/m² BGF. Die Beschriftung zeigt die Über- oder Unterschreitung durch den Gesamtwert.

### Wozu dienen die hinterlegten Bereiche?

Die Bereiche **L1 (0–70)**, **L2 (70–195)** und **L3 (195–320 kg CO₂e/m² BGF)** bieten eine Orientierung innerhalb der vorhandenen Darstellung. Der Budgetvergleich bezieht sich auf die Gesamtbilanz. Für eine abschließende Bewertung müssen Bezugsfläche, Betrachtungszeitraum und Systemgrenze zusammenpassen.
